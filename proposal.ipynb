# Airbnb: Do Listing Details Reflect Guest Experiences?

## Motivation & problem

When people book a property on Airbnb, they depend on a few characteristics to evaluate the listing: overall rating, Superhost status, and amenities provided. However, many listings are rated between 4.7 and 5.0, so the ratings are not a good reflection of what are good stays and which stays were “just okay.” The real opinions and feedback are typically buried in the written reviews.
I’m interested in finding out whether a high rating actually tells you what the stay will be like. This leads me to my research question - do listings rated 4.8+ still have recurring cleanliness and noise complaints in their reviews that the rating doesn’t show?
This is significant to any users booking a place, and this decides what to surface on a listing page. In my usability research class, I am also working on a project that is related to Airbnb, where we are testing users to evaluate the intuitiveness of the product itself. I’m curious to see what types of parallels I will see by evaluating broader data and how the smaller sampling size would compare. I expect to learn how well the 5-star ratings summarizes specific experiences, and where it falls short.

## Data

For this project, I am planning on using Inside Airbnb’s Seattle data (License: CC by 4.0):
- listings.csv: property type, neighborhood, price, host status (including Superhost), overall rating, and category subscores like cleanliness and communication 
- reviews.csv: guest-written reviews with dates, linked to listings by listing ID 
I will use two files because it lets me compare between a listing and what guests wrote. calendar.csv is not relevant to my question, so I will leave that out.

**Ethical considerations:**
Reviews are written by real people and often name hosts. To address this, I will censor reviewer’s names and IDs from anything I save, avoid quoting reviews that could identify someone, and understand that one unfair review shouldn’t define a host’s listing. 

**Plan for license:**
Because the data asks not to be republished, I’ll check with my professor or TA about what the repo needs. My fallback is a download script, my analysis code, and a cleaned, anonymized derived dataset instead of the raw files.
